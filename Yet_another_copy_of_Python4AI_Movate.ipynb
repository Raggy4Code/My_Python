Division is quite interesting in Python, wr we use (/, //, & %)

```python
a = 10
print(a/3)
# 10 / 3 results in 3.333... (Float Division: gives the exact decimal).
```

```python
a = 10
print(a//3)

# 10 // 3 results in 3 (Floor Division: gives the whole number, drops the decimal).
```

```python
a = 10
print(a%3)
# When you divide (10) by (3), it goes in exactly 3 time with one as a remainder. Since the "leftover" value is "1", the result is 1
```
